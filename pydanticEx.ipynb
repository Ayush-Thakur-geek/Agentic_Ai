```python
class PatientData(BaseModel):
    name: str = Field(max_length=50, min_length=5)
    age: int
    weight: float
    height: float
    married: bool
    allergies: Optional[List[str]] = None
    email: EmailStr
    contact_info: Dict[str, str]

    @computed_field
    @property
    def bmi(self) -> float:
        return round(self.weight / (self.height ** 2), 2)

    @field_validator('email')
    @classmethod
    def validate_domain(cls, value):
        valid_domains = ['hdfc.com', 'sbi.com']
        domain_name = value.split('@')[-1]

        if domain_name not in valid_domains:
            raise ValueError("Invalid domain")

        return value

    @field_validator('name')
    @classmethod
    def transform_name_to_upper(cls, value):
        return value.upper()

    @model_validator(mode='after')
    def validate_emergency_contact(self):
        if (
            self.age > 60
            and 'emergency contact' not in self.contact_info
        ):
            raise ValueError("Emergency contact needed")

        return self


def addPatientData(patient: PatientData):
    print(patient.name)
    print(patient.age)
    print(patient.weight)
    print(patient.height)
    print(patient.bmi)
    print(patient.married)
    print(patient.allergies)
    print(patient.contact_info)
    print("Data saved")

patient_data = {"name": "Ayush", "age": "61", "weight": 84, "height": 1.8288, "married": "False", "allergies": ["dust"], "email": "ayush@hdfc.com", "contact_info": {"email": "ayush26819@gmail.com", "phone_no": "7812324234", "emergency contact": "782132421324"}}
patient = PatientData(**patient_data)
patient = PatientData(**patient_data)
addPatientData(patient)
```
