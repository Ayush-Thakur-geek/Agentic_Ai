```python
from pydantic import BaseModel, EmailStr
from typing import List, Dict, Optional
```

```python
class PatientData(BaseModel):
    name: str
    age: int

def addPatientData(patient: PatientData):
    print(patient.name)
    print(patient.age)
    print("Data saved")
patient_data = {"name": "Ayush", "age": "22"}
patient = PatientData(**patient_data)
addPatientData(patient)
```

```python
class PatientData(BaseModel):
    name: str
    age: int
    weight: float
    married: bool
    allergies: Optional[List[str]] = None
    email: EmailStr
    contact_info: Dict[str, str]

def addPatientData(patient: PatientData):
    print(patient.name)
    print(patient.age)
    print(patient.weight)
    print(patient.married)
    print(patient.allergies)
    print(patient.contact_info)
    print("Data saved")
patient_data = {"name": "Ayush", "age": "22", "weight": 84, "married": "False", "allergies": ["dust"], "email": "ayush@gmail.com", "contact_info": {"email": "ayush26819@gmail.com", "phone_no": "7812324234"}}
patient = PatientData(**patient_data)
addPatientData(patient)
```
