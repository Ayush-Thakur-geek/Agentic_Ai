```python
import os
import certifi
import requests

from dotenv import load_dotenv

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_community.tools.tavily_search import TavilySearchResults

from langchain_classic import hub
from langchain_classic.agents import create_react_agent, AgentExecutor

from langchain_core.tools import tool
```

```python
os.environ["SSL_CERT_FILE"] = certifi.where()
load_dotenv()

GENAI_API_KEY = os.getenv("GOOGLE_API_KEY")
TAVILY_API_KEY = os.getenv("TAVILY_API_KEY")
```

```python
search_tool = TavilySearchResults(max_results=5, api_key=TAVILY_API_KEY)
```

```python
search_tool.invoke("what is the reason for the war in Ukraine?")
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite",   # swap for any "flash" name from your list
    google_api_key=os.getenv("GOOGLE_API_KEY"),
    temperature=0.3,
    timeout=60,
    max_retries=2,
)
print(llm.invoke("What is the capital of France?").content)
```

```python
from langsmith import Client

client = Client()

prompt = client.pull_prompt(
    "hwchase17/react",
    dangerously_pull_public_prompt=True
)
```

```python
tools = [search_tool]

agent = create_react_agent(
    llm=llm,
    tools=tools,
    prompt=prompt,
)
```

```python
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True)
```

```python
response = agent_executor.invoke({
    "input": "what is this heavy feeling that we feel while forcing ourselves doing the important work and why does it not go away even when we do not do that work. find some studies related to it"
    })
```
